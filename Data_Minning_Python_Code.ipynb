```python
import csv
import random

first_names = ["Alex", "Jamie", "Taylor", "Jordan", "Casey", "Morgan", "Riley",
               "Cameron", "Dana", "Avery", "Quinn", "Skyler", "Drew", "Charlie", "Harper"]
last_names = ["Smith", "Johnson", "Williams", "Brown", "Jones", "Miller",
              "Davis", "Garcia", "Rodriguez", "Wilson", "Martinez", "Anderson",
              "Thomas", "Hernandez", "Moore"]
genders = ["Male", "Female", "Non-binary"]
cities = ["London", "New York", "Paris", "Tokyo", "Sydney", "Toronto", "Berlin",
          "Madrid", "Seoul", "Cape Town"]
education_levels = ["High School", "Bachelor's", "Master's", "PhD"]


people = []
for i in range(1, 1001):
    first = random.choice(first_names)
    last = random.choice(last_names)
    name = f"{first} {last}"
    gender = random.choice(genders)
    education = random.choice(education_levels)
    age = random.randint(18, 85)
    city = random.choice(cities)
    temperature = round(random.normalvariate(22, 4), 1)  #Interval variable
    duplicate_flag = random.choice([0, 1]) if i % 100 == 0 else 0  # few duplicates
    people.append([i, name, gender, education, age, city, temperature, duplicate_flag])

# Introduce missing city values
for p in random.sample(people, 20):
    p[5] = ""  # missing city

# Write to CSV
with open("people1.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["person_id", "name", "gender", "education_level", "age", "city", "temperature_c", "duplicate_flag"])
    writer.writerows(people)

print("people1.csv")
```

```python
import csv
import random
import datetime

satisfaction_levels = ["Poor", "Average", "Good", "Excellent"]
payment_methods = ["Cash", "Card", "Insurance"]


visits = []
visit_id = 1
for _ in range(1060):
    person_id = random.randint(1, 1000)
    visit_date = (datetime.date(2023, 1, 1) + datetime.timedelta(days=random.randint(0, 600))).isoformat()
    satisfaction = random.choice(satisfaction_levels)
    payment = random.choice(payment_methods)
    cost = round(random.uniform(50, 500), 2)
    notes = random.choice(["", "Follow-up", "Routine check", "Prescription update"])
    visits.append([visit_id, person_id, visit_date, satisfaction, payment, cost, notes])
    visit_id += 1

# Write to CSV
with open("visits1.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["visit_id", "person_id", "visit_date", "satisfaction_level", "payment_method", "cost_usd", "notes"])
    writer.writerows(visits)

print("visits1.csv")
```

```python
import csv
import random

measurements = []

# Simulate 1050 visits (like from visits.csv)
for visit_id in range(1, 1050):
    person_id = random.randint(1, 1000)
    height = round(random.normalvariate(170, 10), 1)
    weight = round(random.normalvariate(70, 15), 1)
    bmi = round(weight / ((height / 100) ** 2), 1)
    body_temp = round(random.normalvariate(36.8, 0.6), 1)
    systolic_bp = random.randint(90, 140)
    measurements.append([person_id, visit_id, height, weight, bmi, body_temp, systolic_bp])

# Write to CSV
with open("measurements1.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["person_id", "visit_id", "height_cm", "weight_kg", "bmi", "body_temp_c", "systolic_bp"])
    writer.writerows(measurements)

print("measurements1.csv")
```
